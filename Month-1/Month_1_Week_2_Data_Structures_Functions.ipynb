## 1. Python Data Structures

```python
# Lists

fruits = ["Apple", "Banana", "Mango", "Orange"]

print("Original List:", fruits)
print("First Fruit:", fruits[0])

fruits.append("Grapes")
print("After Adding:", fruits)

fruits.remove("Banana")
print("After Removing:", fruits)
```

```python
# Sets

numbers = {10, 20, 30, 20, 40, 30}

print("Set:", numbers)

numbers.add(50)
print("After Adding:", numbers)
```

## 2. Functions

Functions are reusable blocks of code designed to perform a specific task.

```python
# Basic Function

def calculate_average(numbers):
    total = sum(numbers)
    count = len(numbers)
    return total / count


temperatures = [28, 30, 29, 31, 27]

average = calculate_average(temperatures)

print("Temperatures:", temperatures)
print("Average Temperature:", average)
```

```python
# Function with Parameters

def calculate_total(price, quantity):
    return price * quantity


price = 250
quantity = 3

total = calculate_total(price, quantity)

print("Price:", price)
print("Quantity:", quantity)
print("Total:", total)
```

## 3. Lambda Functions

Lambda functions are small anonymous functions that can be used for simple
operations and data transformations.

```python
# Lambda Function

square = lambda x: x ** 2

numbers = [1, 2, 3, 4, 5]
squares = [square(number) for number in numbers]

print("Numbers:", numbers)
print("Squares:", squares)
```

```python
# Lambda Function for Filtering

numbers = [10, 15, 20, 25, 30, 35, 40]

even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print("Original Numbers:", numbers)
print("Even Numbers:", even_numbers)
```

## 4. Recursion

Recursion is a technique in which a function calls itself to solve a problem
by breaking it into smaller versions of the same problem.

```python
# Recursive Function – Factorial

def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)


number = 5
result = factorial(number)

print("Number:", number)
print("Factorial:", result)
```

## 5. List Comprehension

List comprehension provides a concise way to create a new list by applying
an expression to each item in an existing iterable.

```python
# List Comprehension

numbers = [1, 2, 3, 4, 5, 6]

squares = [number ** 2 for number in numbers]

print("Original Numbers:", numbers)
print("Squared Numbers:", squares)
```

```python
# List Comprehension with Condition

numbers = [10, 15, 20, 25, 30, 35, 40]

even_numbers = [number for number in numbers if number % 2 == 0]

print("Original Numbers:", numbers)
print("Even Numbers:", even_numbers)
```

## 6. Data Transformation

Data transformation involves modifying or processing data into a useful
form using Python data structures and functions.

```python
# Data Transformation using a Function

def transform_numbers(numbers):
    return [number ** 2 for number in numbers]


numbers = [2, 4, 6, 8, 10]

squared_numbers = transform_numbers(numbers)

print("Original Numbers:", numbers)
print("Squared Numbers:", squared_numbers)
```

```python
# Data Filtering

def filter_above_value(numbers, threshold):
    return [number for number in numbers if number > threshold]


numbers = [10, 25, 15, 40, 30, 5]
threshold = 20

filtered_numbers = filter_above_value(numbers, threshold)

print("Original Numbers:", numbers)
print("Numbers above", threshold, ":", filtered_numbers)
```

## 7. Client Project – Data Cleaning

This project demonstrates basic data cleaning using Python.
The dataset contains duplicate records and records that can be
filtered based on a specified condition.

The cleaning process includes:
- Identifying and removing duplicate records
- Filtering records based on a condition
- Displaying the cleaned data

```python
# Sample Dataset

customer_data = [
    {"name": "Amit", "age": 25, "city": "Kolkata", "purchase": 1200},
    {"name": "Priya", "age": 30, "city": "Delhi", "purchase": 2500},
    {"name": "Rahul", "age": 22, "city": "Mumbai", "purchase": 800},
    {"name": "Sneha", "age": 28, "city": "Kolkata", "purchase": 1800},
    {"name": "Amit", "age": 25, "city": "Kolkata", "purchase": 1200},
    {"name": "Arjun", "age": 35, "city": "Chennai", "purchase": 3200},
    {"name": "Priya", "age": 30, "city": "Delhi", "purchase": 2500}
]

print("Original Dataset:")
for customer in customer_data:
    print(customer)
```

```python
# Remove Duplicate Records

unique_customers = []

for customer in customer_data:
    if customer not in unique_customers:
        unique_customers.append(customer)

print("Dataset after removing duplicates:")
for customer in unique_customers:
    print(customer)
```

```python
# Filter Records

minimum_purchase = 1500

filtered_customers = [
    customer
    for customer in unique_customers
    if customer["purchase"] > minimum_purchase
]

print("Customers with purchases above ₹1500:")
for customer in filtered_customers:
    print(customer)
```

```python
# Data Cleaning Function

def clean_customer_data(data, minimum_purchase):
    unique_data = []

    # Remove duplicates
    for record in data:
        if record not in unique_data:
            unique_data.append(record)

    # Filter records
    cleaned_data = [
        record
        for record in unique_data
        if record["purchase"] > minimum_purchase
    ]

    return cleaned_data


cleaned_customers = clean_customer_data(customer_data, 1500)

print("Final Cleaned Dataset:")
for customer in cleaned_customers:
    print(customer)
```
