## 1. Libraries and Dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully.")
```

```python
# Create Dataset

data = {
    "Customer": ["Amit", "Priya", "Rahul", "Sneha", "Arjun",
                 "Neha", "Rohan", "Anjali", "Vikram", "Pooja"],
    "Age": [25, 30, 22, 28, 35, 27, 29, 24, 32, 26],
    "City": ["Kolkata", "Delhi", "Mumbai", "Kolkata", "Chennai",
             "Delhi", "Mumbai", "Kolkata", "Chennai", "Delhi"],
    "Purchase": [1200, 2500, 800, 1800, 3200,
                 2100, 1500, 1100, 2800, 2200],
    "Visits": [3, 5, 2, 4, 6, 5, 3, 2, 6, 4],
    "Satisfaction": [3, 5, 2, 4, 5, 4, 3, 3, 5, 4]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Basic Dataset Information

print("Dataset Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nSummary Statistics:")
print(df.describe())
```

## 2. Matplotlib Visualizations

### 2.1 Histogram – Purchase Distribution

```python
# Histogram of Purchase Amounts

plt.figure(figsize=(8, 5))

plt.hist(df["Purchase"], bins=5, edgecolor="black")

plt.title("Distribution of Purchase Amounts")
plt.xlabel("Purchase Amount")
plt.ylabel("Number of Customers")

plt.show()
```

### 2.2 Bar Chart – Average Purchase by City

```python
# Average Purchase by City

city_purchase = df.groupby("City")["Purchase"].mean()

plt.figure(figsize=(8, 5))

plt.bar(city_purchase.index, city_purchase.values)

plt.title("Average Purchase by City")
plt.xlabel("City")
plt.ylabel("Average Purchase")

plt.xticks(rotation=30)

plt.show()
```

### 2.3 Scatter Plot – Visits vs Purchase

```python
# Scatter Plot

plt.figure(figsize=(8, 5))

plt.scatter(df["Visits"], df["Purchase"])

plt.title("Visits vs Purchase")
plt.xlabel("Number of Visits")
plt.ylabel("Purchase Amount")

plt.show()
```

### 2.4 Line Plot – Purchase by Customer

```python
# Line Plot

plt.figure(figsize=(10, 5))

plt.plot(df["Customer"], df["Purchase"], marker="o")

plt.title("Purchase Amount by Customer")
plt.xlabel("Customer")
plt.ylabel("Purchase Amount")

plt.xticks(rotation=45)

plt.show()
```

### 3.1 Seaborn Bar Plot – Average Purchase by City

```python
# Seaborn Bar Plot

plt.figure(figsize=(8, 5))

sns.barplot(data=df, x="City", y="Purchase")

plt.title("Average Purchase by City")
plt.xlabel("City")
plt.ylabel("Average Purchase")

plt.xticks(rotation=30)

plt.show()
```

### 3.2 Box Plot – Purchase Distribution

```python
# Box Plot

plt.figure(figsize=(8, 5))

sns.boxplot(data=df, y="Purchase")

plt.title("Purchase Amount Distribution")
plt.ylabel("Purchase Amount")

plt.show()
```

### 3.3 Heatmap – Feature Correlation

```python
# Correlation Heatmap

plt.figure(figsize=(8, 6))

correlation = df[["Age", "Purchase", "Visits", "Satisfaction"]].corr()

sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Feature Correlation Heatmap")

plt.show()
```

### 3.4 Pairplot – Relationships Between Features

```python
# Pairplot

sns.pairplot(
    df[["Age", "Purchase", "Visits", "Satisfaction"]]
)

plt.show()
```

## 4. Client Project – Data Visualization Dashboard

This project visualizes relationships and patterns in customer data
using Matplotlib and Seaborn.

The dashboard includes:
- Purchase distribution
- Average purchase by city
- Visits versus purchase relationship
- Feature correlation
- Relationships between numerical features

```python
# Dashboard Visualization 1: Purchase Distribution

plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="Purchase", bins=5, kde=True)

plt.title("Purchase Amount Distribution")
plt.xlabel("Purchase Amount")
plt.ylabel("Number of Customers")

plt.show()
```

```python
# Dashboard Visualization 2: Average Purchase by City

city_purchase = df.groupby("City")["Purchase"].mean()

plt.figure(figsize=(8, 5))

plt.bar(city_purchase.index, city_purchase.values)

plt.title("Average Purchase by City")
plt.xlabel("City")
plt.ylabel("Average Purchase")

plt.xticks(rotation=30)

plt.show()
```

```python
# Dashboard Visualization 3: Visits vs Purchase

plt.figure(figsize=(8, 5))

sns.scatterplot(data=df, x="Visits", y="Purchase", s=100)

plt.title("Visits vs Purchase")
plt.xlabel("Number of Visits")
plt.ylabel("Purchase Amount")

plt.show()
```

```python
# Dashboard Visualization 4: Correlation Heatmap

correlation = df[["Age", "Purchase", "Visits", "Satisfaction"]].corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Feature Correlation Heatmap")

plt.show()
```

```python
# Dashboard Visualization 5: Feature Relationships

sns.pairplot(
    df[["Age", "Purchase", "Visits", "Satisfaction"]]
)

plt.show()
```
