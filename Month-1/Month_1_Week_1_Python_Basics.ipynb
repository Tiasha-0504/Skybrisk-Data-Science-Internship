## 1. Python Programming Basics

This section demonstrates the fundamental concepts of Python programming, including variables, data types, operators, input/output, conditional statements, and loops.

```python
# Variables and Data Types

name = "Tiasha"
age = 22
cgpa = 8.37
is_student = True

print("Name:", name)
print("Age:", age)
print("CGPA:", cgpa)
print("Student:", is_student)

print("\nData Types:")
print(type(name))
print(type(age))
print(type(cgpa))
print(type(is_student))
```

## 2. Operators in Python

Python provides different types of operators for performing calculations,
comparisons, and logical operations.

```python
# Arithmetic Operators

a = 20
b = 6

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Exponentiation:", a ** b)
```

```python
# Comparison and Logical Operators

x = 10
y = 5

print("x > y:", x > y)
print("x < y:", x < y)
print("x == y:", x == y)
print("x != y:", x != y)

print("Logical AND:", x > 5 and y < 10)
print("Logical OR:", x < 5 or y < 10)
print("Logical NOT:", not (x < y))
```

## 3. Input and Output

The input() function is used to accept data from the user, while print()
is used to display information.

```python
# User Input and Output

name = input("Enter your name: ")
age = int(input("Enter your age: "))

print("\nHello,", name)
print("You are", age, "years old.")
```

## 4. Conditional Statements

Conditional statements allow a program to execute different blocks of code
depending on whether a condition is true or false.

```python
# If-Else Example

marks = 75

if marks >= 40:
    print("Result: Pass")
else:
    print("Result: Fail")
```

```python
# If-Elif-Else Example

marks = 75

if marks >= 90:
    grade = "A+"
elif marks >= 80:
    grade = "A"
elif marks >= 70:
    grade = "B"
elif marks >= 60:
    grade = "C"
else:
    grade = "D"

print("Grade:", grade)
```

### For Loop

A for loop is used to iterate over a sequence of values.

```python
# For Loop Example

for number in range(1, 6):
    print("Number:", number)
```

### While Loop

A while loop repeatedly executes a block of code while a condition remains true.

```python
# While Loop Example

count = 1

while count <= 5:
    print("Count:", count)
    count += 1
```

## 5. Hands-on Programs

### 5.1 Calculator

```python
# Basic Calculator

num1 = float(input("Enter the first number: "))
operator = input("Enter an operator (+, -, *, /): ")
num2 = float(input("Enter the second number: "))

if operator == "+":
    result = num1 + num2
elif operator == "-":
    result = num1 - num2
elif operator == "*":
    result = num1 * num2
elif operator == "/":
    if num2 != 0:
        result = num1 / num2
    else:
        result = "Error: Division by zero is not allowed."
else:
    result = "Error: Invalid operator."

print("Result:", result)
```

### 5.2 Temperature Converter

```python
# Temperature Converter

temperature = float(input("Enter temperature: "))
unit = input("Enter the unit (C for Celsius, F for Fahrenheit): ").upper()

if unit == "C":
    converted_temperature = (temperature * 9 / 5) + 32
    print(f"{temperature}°C = {converted_temperature:.2f}°F")

elif unit == "F":
    converted_temperature = (temperature - 32) * 5 / 9
    print(f"{temperature}°F = {converted_temperature:.2f}°C")

else:
    print("Invalid unit. Please enter C or F.")
```

## 6. Client Project – Temperature Data Processing

This project demonstrates basic data processing using Python.
The program accepts a set of temperature values and calculates
the average temperature.

```python
# Temperature Data Processing

temperatures = [28.5, 30.2, 29.8, 31.5, 27.9, 30.6, 29.4]

total_temperature = sum(temperatures)
number_of_readings = len(temperatures)
average_temperature = total_temperature / number_of_readings

print("Temperature Readings:", temperatures)
print("Number of Readings:", number_of_readings)
print("Average Temperature:", round(average_temperature, 2), "°C")
```
