## 1. NumPy Arrays and Operations

NumPy is a Python library used for numerical computing and efficient
operations on arrays.

```python
import numpy as np

# Creating a NumPy Array

temperatures = np.array([28.5, 30.2, 29.8, 31.5, 27.9])

print("Temperature Array:", temperatures)
print("Data Type:", temperatures.dtype)
print("Number of Elements:", temperatures.size)
```

```python
# NumPy Arithmetic Operations

temperatures = np.array([28, 30, 32, 29, 31])

print("Original Temperatures:", temperatures)
print("After Adding 2:", temperatures + 2)
print("After Subtracting 2:", temperatures - 2)
print("Doubled Values:", temperatures * 2)
print("Half Values:", temperatures / 2)
```

```python
# NumPy Aggregate Operations

temperatures = np.array([28, 30, 32, 29, 31])

print("Mean:", np.mean(temperatures))
print("Maximum:", np.max(temperatures))
print("Minimum:", np.min(temperatures))
print("Total:", np.sum(temperatures))
```

```python
# NumPy Indexing and Slicing

temperatures = np.array([28, 30, 32, 29, 31])

print("First Temperature:", temperatures[0])
print("Last Temperature:", temperatures[-1])
print("First Three Temperatures:", temperatures[:3])
print("Temperatures from Index 2:", temperatures[2:])
```

### Broadcasting

Broadcasting allows NumPy to perform arithmetic operations between
an array and a single value without explicitly repeating that value.

```python
# NumPy Broadcasting

temperatures = np.array([28, 30, 32, 29, 31])

adjusted_temperatures = temperatures + 1.5

print("Original Temperatures:", temperatures)
print("Adjusted Temperatures:", adjusted_temperatures)
```

## 2. Pandas Series and DataFrames

Pandas provides data structures such as Series and DataFrames for
working with and manipulating structured data.

```python
import pandas as pd

# Creating a Pandas Series

temperatures = pd.Series([28.5, 30.2, 29.8, 31.5, 27.9])

print("Temperature Series:")
print(temperatures)
```

```python
# Series Operations

print("Mean Temperature:", temperatures.mean())
print("Maximum Temperature:", temperatures.max())
print("Minimum Temperature:", temperatures.min())
```

```python
# Creating a Pandas DataFrame

data = {
    "Name": ["Amit", "Priya", "Rahul", "Sneha", "Arjun"],
    "City": ["Kolkata", "Delhi", "Mumbai", "Kolkata", "Chennai"],
    "Age": [25, 30, 22, 28, 35],
    "Purchase": [1200, 2500, 800, 1800, 3200]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Basic DataFrame Information

print("First 3 Rows:")
print(df.head(3))

print("\nDataFrame Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)
```

```python
# DataFrame Indexing

print("Purchase Column:")
print(df["Purchase"])

print("\nFirst Row:")
print(df.iloc[0])

print("\nFirst Three Rows:")
print(df.iloc[:3])
```

```python
# Filtering Data

high_value_customers = df[df["Purchase"] > 1500]

print("Customers with Purchase > 1500:")
print(high_value_customers)
```

```python
# Grouping Data

city_purchase = df.groupby("City")["Purchase"].mean()

print("Average Purchase by City:")
print(city_purchase)
```

## 3. Client Project – Dataset Cleaning and Aggregation

This project demonstrates data cleaning and aggregation using Pandas.

The dataset contains customer purchase information with some missing values.
The cleaning process removes incomplete records, and the cleaned data is
then aggregated to calculate average purchase values.

```python
# Create Sample Dataset

data = {
    "Customer": ["Amit", "Priya", "Rahul", "Sneha", "Arjun", "Neha", "Rohan"],
    "City": ["Kolkata", "Delhi", "Mumbai", "Kolkata", "Chennai", "Delhi", "Mumbai"],
    "Age": [25, 30, 22, None, 35, 27, 29],
    "Purchase": [1200, 2500, None, 1800, 3200, 2100, 1500]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)
```

```python
# Check for Missing Values

print("Missing Values:")
print(df.isnull().sum())
```

```python
# Remove Rows with Missing Values

cleaned_df = df.dropna()

print("Dataset After Removing Missing Values:")
print(cleaned_df)
```

```python
# Calculate Average Purchase

average_purchase = cleaned_df["Purchase"].mean()

print("Average Purchase:", round(average_purchase, 2))
```

```python
# Average Purchase by City

city_summary = (
    cleaned_df
    .groupby("City")["Purchase"]
    .mean()
    .reset_index()
)

print("Average Purchase by City:")
print(city_summary)
```

```python
# Final Cleaned Dataset

print("Number of Original Records:", len(df))
print("Number of Cleaned Records:", len(cleaned_df))

print("\nFinal Cleaned Dataset:")
print(cleaned_df)
```
